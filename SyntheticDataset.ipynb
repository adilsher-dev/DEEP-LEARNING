```python

# Example 2 - House Price Prediction (Regression) with Synthetic Dataset

import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np

np.random.seed(42)

X_train = np.random.rand(1000, 10)
y_train = np.random.rand(1000) * 500000  # Random prices between 0 and 500K

X_test = np.random.rand(200, 10)
y_test = np.random.rand(200) * 500000

# Build model
model = models.Sequential([
    layers.Dense(64, activation='relu', input_shape=(10,)),
    layers.Dense(32, activation='relu'),
    layers.Dense(1)  # Single output for regression
])

# Compile model
model.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mae']
)

# Train model
model.fit(
    X_train, y_train,
    epochs=20,
    batch_size=32,
    verbose=1
)

# Evaluate model
test_loss, test_mae = model.evaluate(X_test, y_test)

print(f'Test MAE: ${test_mae:,.2f}')

# Make a prediction
sample_input = np.random.rand(1, 10)

predicted_price = model.predict(sample_input, verbose=0)
print(f'Predicted House Price: ${predicted_price[0][0]:,.2f}')
```
