```python

# Example 2 - House Price Prediction (Regression)
# Synthetic Dataset and Normalization

import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np

np.random.seed(42)

X_train = np.random.rand(1000, 10)
y_train = np.random.rand(1000)  # Target values between 0 and 1

X_test = np.random.rand(200, 10)
y_test = np.random.rand(200)    # Target values between 0 and 1


model = models.Sequential([
    layers.Input(shape=(10,)),
    layers.Dense(64, activation='relu'),
    layers.Dense(32, activation='relu'),
    layers.Dense(1)  # Single output for regression
])


model.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mae']
)


model.fit(
    X_train, y_train,
    epochs=20,
    batch_size=32,
    verbose=1
)


test_loss, test_mae = model.evaluate(X_test, y_test)

print(f'Test MAE (Normalized): {test_mae:.4f}')


sample_input = np.random.rand(1, 10)
predicted_value = model.predict(sample_input, verbose=0)

print(f'Predicted Normalized Value: {predicted_value[0][0]:.4f}')
```
